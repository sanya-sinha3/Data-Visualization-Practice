```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
scores=np.random.normal(70,10,100) #YE APNE APP 100 VALUES KO LE LEGA JISKA KI MEAN 70 STANDARD DEVIATION 10 HOGA
plt.hist(scores,color="blue",edgecolor="black",bins=10)
plt.ylabel("no. of students")
plt.xlabel("scores")
```

```python
#DIFFERENT WAYS TO WRITE BINS

np.random.seed(0)
scores=np.random.normal(70,10,100)
bins=[0,30,70,90,100]
plt.hist(scores,color="blue",edgecolor="black",bins=bins)
plt.ylabel("no. of students")
plt.xlabel("scores")
```

```python
#MULTIPLE PLOTTING

legit_transaction=[2.99,5.49,8.99,12.50,14.99,19.99,23.45,29.99,34.99,39.50,44.99,49.99,55.25,60.00,75.99,89.99,120.50,149.99,199.99,250.00,300.75,450,600,850,1200]
fraud_transaction=[50,100,150,200,300,500,500,750,1000,1000,1200,1500,1500,2000,2500,3000,3000]
plt.hist(fraud_transaction,color="red",bins=10,label="fraud" ,edgecolor="black",alpha=0.5)
plt.hist(legit_transaction,color="green",bins=10,label="legit",edgecolor="black",alpha=0.5)
plt.title("legit vs fraud")
plt.xlabel("amount")
plt.ylabel("frequency")
plt.legend()
```

```python
#AXVLINE

plt.axvline(
    x=33,label="passing marks",color="red",linewidth="2",linestyle="--")
```

```python
#PRACTICE QUESTION

np.random.seed(0)
scores=np.random.normal(70,10,200) 
mean_val = np.mean(scores)

plt.hist(scores,color="blue",edgecolor="black",bins=10)
plt.ylabel("no. of students")
plt.xlabel("scores")

plt.axvline(
    x=mean_val,label="mean",color="red",linewidth="2",linestyle="--")
```
