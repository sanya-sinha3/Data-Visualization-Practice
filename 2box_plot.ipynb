```python
import matplotlib.pyplot as plt
```

```python
#VERTICLE BAR CAHRTS

oscar_movies=["the dark knoghts","the hurt locker","argo","the kings speech","the artist"]
years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,425,133,232]
plt.bar(years,oscar_revenue)
plt.title("oscar movies revenue in each year")  #TITLE METHOD
plt.xlabel("years")      #XLABEL METHOD
plt.ylabel("revenue (in $M)")  #YLABEL METHOD
```

```python
#METHODS

oscar_movies=["the dark knoghts","the hurt locker","argo","the kings speech","the artist"]
years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,425,133,232]
plt.bar(years,oscar_revenue)
plt.title("oscar movies revenue in each year")  #TITLE METHOD
plt.xlabel("years")      #XLABEL METHOD
plt.ylabel("revenue (in $M)")  #YLABEL METHOD

for i in range(len(years)):
  plt.text(years[i],oscar_revenue[i]+5,str(oscar_revenue[i]),ha="center") #BOX KE UPAR TEXT LIKHNE KE LIYE

#GRAPH KI SIZE BADHANE KE LIYE
plt.ylim(0,max(oscar_revenue)+1)
plt.xlim(min(years),max(years)+1)
```

```python
#COMPARE 2 SETS OF DATA

import numpy as np
oscar_movies=["the dark knoghts","the hurt locker","argo","the kings speech","the artist"]
years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,425,133,232]
non_oscar_movies=["slumdag","avatar","inception","hugo","linden"]
non_oscar_revenue=[378,2788,829,185,275]

#x is the position of each category on the x axis
x=np.arange(len(years)) #5 YEARS VALUE KO ARRAY ME CONVERT KAREGA (0,1,2,3,4) AB X AXIS PR YAERA KE JAGAH 01234 SHOW KAREGA
width=0.4 #each bar width

plt.bar(x-width/2,oscar_revenue,width,label="oscar") #OSCAR KO LEFT ME SHIFT KR RHE
plt.bar(x+width/2,oscar_revenue,width,label="non-oscar") #NON-OSCAR KO RIGHT ME SHIFT KR RHE

plt.title("oscar vs non-oscar")  #TITLE METHOD
plt.xlabel("years")      #XLABEL METHOD
plt.ylabel("revenue (in $M)")  #YLABEL METHOD
plt.legend()

plt.xticks(x,years)  #X KI VALUE KO YEARS ME EXCAHNGE KE DENGE
```
