```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=[7,8,5,6,9,4,10,12,15]
plt.boxplot(data)
plt.grid(True)
```

```python
#OPERATION

#MULTIPLE DATASETS
grp1=np.random.normal(50,10,100)
grp2=np.random.normal(60,15,100)
plt.boxplot([grp1,grp2],labels=["group1","group2"])
plt.show()
```

```python
#HORIZONTAL BOX PLOT

data=[7,8,5,6,9,4,10,12,15]
plt.boxplot(data,vert=False)

#SHOW MEAN
plt.boxplot(data,vert=False,showmeans=True)

#CHANGE WHISKER LENGTH
plt.boxplot(data,vert=False,showmeans=True,whis=4.0)
```
