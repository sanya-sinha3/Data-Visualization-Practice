```python
import matplotlib.pyplot as plt
```

```python
people=["A","B","C","D","E","F","G"]
age=[22,25,30,35,40,45,50]
bp=[110,115,120,170,130,135,155]

plt.scatter(age,bp)
plt.title("age vs bp")
plt.xlabel("age")
plt.ylabel("bp")
```

```python
#CUSTOMIATION

people=["A","B","C","D","E","F","G"]
age=[22,25,30,35,40,45,50]
bp=[110,115,120,170,130,135,155]

color=["green" if x<135 else "red" for x in bp]

plt.scatter(age,bp,c=color,s=bp,marker="o",alpha=0.5)
plt.title("age vs bp")
plt.xlabel("age")
plt.ylabel("bp")
```

```python
#COLORMAP AND COLOR BAR

people=["A","B","C","D","E","F","G"]
age=[22,25,30,35,40,45,50]
bp=[110,115,120,170,130,135,155]

plt.scatter(age,bp,cmap="plasma",c=bp)
plt.title("age vs bp")
plt.xlabel("age")
plt.ylabel("bp")
plt.colorbar(label="bp")
```

```python
#ANNOTATION

people=["A","B","C","D","E","F","G"]
age=[22,25,30,35,40,45,50]
bp=[110,115,120,170,130,135,155]

plt.scatter(age,bp)
plt.title("age vs bp")
plt.xlabel("age")
plt.ylabel("bp")
for i in range(len(people)):
    plt.annotate(people[i],xy=(age[i],bp[i]),xytext=(age[i]+1,bp[i]+1))
```
