```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4]
y=[5,6,7,8]
plt.plot(x,y) #LINE PLOT
```

```python
#REAL DATA

#OSCAR WINNER MOVIES
oscar_movies=["the dark knoghts","the hurt locker","argo","the kings speech","the artist"]
years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,425,133,232]
plt.plot(years,oscar_revenue)
plt.title("oscar movies revenue in each year")  #TITLE METHOD
plt.xlabel("years")      #XLABEL METHOD
plt.ylabel("revenue (in $M)")  #YLABEL METHOD
```

```python
#OSCAR VS NON OSCAR MOVIES 
#NON-OSCAR WINNER MOVIES

non_oscar_movies=["slumdag","avatar","inception","hugo","linden"]
non_oscar_revenue=[378,2788,829,185,275]
plt.plot(years,oscar_revenue,label="oscar movies")
plt.plot(years,non_oscar_revenue,label="non-oscar movies")
plt.title("oscar movies vs non oscar movies revenue")
plt.xlabel("years")      #XLABEL METHOD
plt.ylabel("revenue (in $M)")
plt.legend()
```

```python
#FORMAT STRINGS

non_oscar_movies=["slumdag","avatar","inception","hugo","linden"]
non_oscar_revenue=[378,2788,829,185,275]
plt.plot(years,oscar_revenue,"o--r",label="oscar movies")
plt.plot(years, non_oscar_revenue, color="red", marker=">", linestyle="-.", linewidth=5, label="non-oscar movies")
plt.title("oscar movies vs non oscar movies revenue")
plt.xlabel("years")      #XLABEL METHOD
plt.ylabel("revenue (in $M)")
plt.legend()
```

```python
#OTHER METHOD AND ATTRIBUTES

non_oscar_movies=["slumdag","avatar","inception","hugo","linden"]
non_oscar_revenue=[378,2788,829,185,275]
plt.plot(years,oscar_revenue,"o--r",label="oscar movies")
plt.plot(years, non_oscar_revenue, color="red", marker=">", linestyle="-.", linewidth=5, label="non-oscar movies")
plt.title("oscar movies vs non oscar movies revenue")
plt.xlabel("years")      #XLABEL METHOD
plt.ylabel("revenue (in $M)")
plt.legend()
plt.style.available
plt.style.use("dark_background")
plt.grid()
plt.tight_layout()
plt.savefig("abc.png")
```
